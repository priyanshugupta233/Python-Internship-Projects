```python
import random

heads = 0
trials = 1000

for i in range(trials):
  if random.choice(["H", "T"]) == "H":
    heads += 1

print("Probability of Head:", heads / trials)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
import numpy as np
import matplotlib.pyplot as plt

# Data
X = np.array([1,2,3,4,5]).reshape(-1,1)
y = np.array([1,4,9,16,25])  # curve data

# Simple model (High Bias)
model = LinearRegression()
model.fit(X, y)

plt.scatter(X, y)
plt.plot(X, model.predict(X))
plt.title("Underfitting (High Bias)")
plt.show()
```

```python
import pandas as pd

data = {
    "Hours_Study": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 90]
}

df = pd.DataFrame(data)
print(df)
print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Std Dev:", df["Marks"].std())
print(df.corr())

import matplotlib.pyplot as plt

plt.scatter(df["Hours_Study"], df["Marks"])
plt.xlabel("Hours Studies")
plt.ylabel("Marks")
plt.title("Study vs Marks")
plt.show()

# Add a student who studied less but scored high
df.loc[len(df)] = [2, 85]
```
